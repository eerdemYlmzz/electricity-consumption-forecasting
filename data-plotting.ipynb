```python

import pandas as pd
import numpy as np
import math
import random
import time
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet("data\processed_powerconsumption.parquet")

print(df.tail())
```

```python
df.shape
```

```python
df.loc[0:5]
```

```python
"""
DATE = YEAR-MONTH-DAY
"""

start_date = "2025-07-01 00:00:00"
df['datetime'] = pd.date_range(start=start_date, periods=len(df), freq='10min') # 10 minute frequency.

plt.figure(figsize=(14, 6))

plt.plot(df['datetime'], df['consumption'], label='Consumption (kWh)', color='orange', linewidth=0.8)

plt.title('Electricity Consumption (Starting July)')
plt.xlabel('Date')
plt.ylabel('kWh')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5) 
plt.tight_layout()
plt.show()
```

```python
start_date = "2025-07-01 00:00:00"

df['datetime'] = pd.date_range(start=start_date, periods=len(df), freq='10min')

df_indexed = df.set_index('datetime')

df_two_days = df_indexed.loc['2025-07-01':'2025-07-05']

plots_dir = Path('plots')
plots_dir.mkdir(exist_ok=True)

plt.figure(figsize=(12, 5))

plt.plot(df_two_days.index, df_two_days['consumption'], label='Daily Consumption (kWh)', color='orange', marker='o', markersize=2)

plt.title('Electricity Consumption')
plt.xlabel('Time')
plt.ylabel('kWh')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

#plt.savefig(plots_dir / 'electricity_consumption_1_5july.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
df_indexed.to_csv('data/data_processed.csv', index=True)
```
